using  set() function of seaborn see how this plot changing 

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns 
df = sns.load_dataset('car_crashes') 
df.head() 
plt.scatter(df.speeding,df.alcohol) 
sns.set() 
plt.show() 
```

Darkgrid 

```python
df = sns.load_dataset('car_crashes') 
df.head() 
plt.scatter(df.speeding,df.alcohol) 
sns.set_style("dark") 
plt.show()
```

We can remove the top and right axis spines using the despine() function.

```python
df = sns.load_dataset('car_crashes') 
df.head() 
plt.scatter(df.speeding,df.alcohol) 
sns.set_style("ticks") 
sns.despine() 
plt.show()
```

using this parameter of axis style to check 

```python
df = sns.load_dataset('car_crashes')
print(df.head())
plt.scatter(df.speeding,df.alcohol) 
sns.set_style("darkgrid",{'grid.color': '.5'}) 
sns.despine() 
plt.show() 
```

By default, context is set to notebook 

```python
df = sns.load_dataset('car_crashes')
print(df.head(10))

sns.set_style("dark")
sns.set_context("notebook")

plt.scatter(df.speeding, df.alcohol)

plt.xlabel("Speeding")
plt.ylabel("Alcohol")
plt.title("Speeding vs Alcohol")

plt.show()
 
```
